Task 1.2: Train-wise Table Showing Route Start and End Stations

Create Route table

```python
route_table = (
    df.groupby('Train_No')
      .apply(lambda x: pd.Series({
          'Start_Station': x.iloc[0]['Station_Name'],
          'End_Station': x.iloc[-1]['Station_Name'],
          'Number_of_Stops': len(x)
      }))
      .reset_index()
)
```
