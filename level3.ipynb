Level 3: Data Exploration

Import Libraries

```python
import pandas as pd
import numpy as np
```

Load Dataset

```python
df = pd.read_csv("Dataset1.csv")
# Level 2 output
final_features = pd.read_csv("final_features.csv")
```

First 5 rows

```python
df.head()
```

Task 3.1: Compare Journey Duration Across Different Routes

```python

route_analysis = (
    final_features.groupby('Train_No')['Journey_Duration_Hours']
    .mean()
    .reset_index()
)

route_analysis.head()
```

Find Longest Routes

```python
route_analysis.sort_values(
    by='Journey_Duration_Hours',
    ascending=False
).head(10)
```

Task 3.2: Identify Stations with High and Low Train Traffic

```python
station_traffic = (
    df.groupby('Station_Name')
      .size()
      .reset_index(name='Train_Count')
)

station_traffic.head()
```

Top 10 Busy Stations

```python
top_stations = station_traffic.sort_values(
    by='Train_Count',
    ascending=False
)

top_stations.head(10)
```

Lowest Traffic Stations

```python
top_stations.tail(10)
```

Task 3.3: Analyze Relationship Between Distance and Journey Duration

```python
correlation = final_features[
    ['Total_Distance', 'Journey_Duration_Hours']
].corr()

print(correlation)
```

Task 3.4: Summarize Key Exploratory Insights

Key Findings
1. Distance vs Duration
Journey duration increases as distance increases.
Strong positive relationship observed.

2. Stops vs Duration

```python
final_features[
    ['Number_of_Stops','Journey_Duration_Hours']
].corr()
```

3. Station Traffic
Large junction stations handle the highest train traffic.
Smaller stations experience lower train frequency.

4. Route Characteristics
Short-distance routes have shorter travel times.
Long-distance routes tend to have more stops and longer durations.

Conclusion

The exploratory analysis revealed meaningful relationships between route characteristics and journey duration. Distance and number of stops both contribute significantly to travel time. Station traffic analysis highlighted major railway hubs and low-traffic stations. These insights will support visualization and machine learning modeling in the next stages.
