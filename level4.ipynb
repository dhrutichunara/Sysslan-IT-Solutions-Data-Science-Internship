Level 4 : Visualization & Pattern Analysis

Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Load Dataset

```python
df = pd.read_csv("Dataset1.csv")
# Level 2 output
final_features = pd.read_csv("final_features.csv")
```

First 5 rows

```python
df.head()
```

Task 4.1: Visualize Journey Duration Across Different Routes

```python
top_routes = final_features.nlargest(
    10,
    'Journey_Duration_Hours'
)

plt.figure(figsize=(10,5))
plt.bar(
    top_routes['Train_No'].astype(str),
    top_routes['Journey_Duration_Hours']
)

plt.title("Top 10 Longest Train Journeys")
plt.xlabel("Train Number")
plt.ylabel("Journey Duration (Hours)")
plt.xticks(rotation=45)

plt.show()
```

Task 4.2: Create Charts Representing Station-wise Train Traffic

```python
station_traffic = (
    df.groupby('Station_Name')
      .size()
      .reset_index(name='Train_Count')
)

top_stations = station_traffic.nlargest(
    10,
    'Train_Count'
)

plt.figure(figsize=(12,5))

plt.bar(
    top_stations['Station_Name'],
    top_stations['Train_Count']
)

plt.title("Top 10 Busy Railway Stations")
plt.xlabel("Station Name")
plt.ylabel("Number of Trains")
plt.xticks(rotation=90)

plt.show()
```

Task 4.3: Plot Distance Versus Journey Duration

```python
plt.figure(figsize=(8,5))

plt.scatter(
    final_features['Total_Distance'],
    final_features['Journey_Duration_Hours']
)

plt.title("Distance vs Journey Duration")
plt.xlabel("Total Distance")
plt.ylabel("Journey Duration (Hours)")

plt.show()
```

Task 4.4: Explain Patterns Observed from Visual Analysis

Pattern 1: Distance vs Duration
Longer distances generally require more travel time.
Scatter plot shows a positive trend.
Pattern 2: Station Traffic
A few stations handle very high traffic.
Most stations receive moderate or low traffic.
Pattern 3: Route Characteristics
Long routes often include more stops.
More stops contribute to increased journey duration.
Pattern 4: Journey Distribution
Some trains are outliers with exceptionally long travel times.
Majority of trains operate within a moderate duration range.

Conclusion

The visual analysis revealed clear patterns in train operations. Journey duration is strongly influenced by route distance and the number of stops. Station traffic analysis identified major railway hubs with high train frequency. These visual insights support the development of predictive models in the next stage.
