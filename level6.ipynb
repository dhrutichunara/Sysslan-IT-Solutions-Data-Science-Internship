Level 6: Final Data Science Project

The objective of this level is to integrate the machine learning model into a complete journey time prediction system and visualize the comparison between actual and predicted journey durations.

Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

Load Dataset

```python
final_features = pd.read_csv("final_features.csv")
```

Task 6.1: Develop Journey Time Prediction System

Prepare Features and Target

```python
X = final_features[['Total_Distance', 'Number_of_Stops']]
y = final_features['Journey_Duration_Hours']
```

Split Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

Train Model

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

Predict Journey Duration

```python
y_pred = model.predict(X_test)
```

Prediction Table

```python
prediction = pd.DataFrame({
    'Actual Journey Duration': y_test.values,
    'Predicted Journey Duration': y_pred
})

prediction.head(10)
```

Visualization: Actual vs Predicted Journey Duration

```python
plt.figure(figsize=(10,5))

plt.plot(y_test.values[:50], label="Actual")

plt.plot(y_pred[:50], label="Predicted")

plt.title("Actual vs Predicted Journey Duration")

plt.xlabel("Test Samples")

plt.ylabel("Journey Duration (Hours)")

plt.legend()

plt.show()
```

Conclusion

A complete train journey time prediction system was successfully developed using the Linear Regression algorithm. The model was trained using Total Distance and Number of Stops as input features and was able to predict journey duration with good accuracy. The comparison between actual and predicted values demonstrates that the model performs effectively and can be used as a basic journey time prediction system
